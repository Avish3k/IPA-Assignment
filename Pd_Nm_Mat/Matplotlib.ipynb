```python
x=np.arange(1,11)
y=np.sin(x)
z=np.cos(x)
print(x,y,z, sep="\n\n")
```

```python
plt.plot(x,y)
#sinusoidal wave
```

```python
#plt.legend gives an example line color for reference 
plt.plot(x,y,label="SINQ", color="red")
plt.plot(x,z,label="COSQ",color="green")
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate values")
plt.xticks()
plt.yticks()
plt.legend()
plt.show()
```

```python
#plt.figure changes the size 8 being the width and 4 being the height
plt.figure(figsize=(8,4))
plt.plot(x,y,label="SINQ", color="red")
plt.plot(x,z,label="COSQ",color="green")
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate values")
plt.xticks(y)
plt.yticks(z)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.plot(x,y,label="SINQ", color="red")
plt.plot(x,z,label="COSQ",color="green")
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate values")
plt.legend()



plt.subplot(1,2,2)
plt.plot(x,y,label="SINQ", color="green")
plt.plot(x,z,label="COSQ",color="red")
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate values")


plt.legend()
plt.show()
```

```python
plt.figure(figsize=(13,7))

plt.subplot(3,3,1)
plt.plot(x,y,label="SINQ", color="red")
plt.plot(x,z,label="COSQ",color="green")
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate values")
plt.legend(loc="upper left")



plt.subplot(3,3,5)
plt.plot(x,y,label="SINQ", color="green")
plt.plot(x,z,label="COSQ",color="red")
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate values")
plt.legend(loc="upper left")

plt.subplot(3,3,9)
plt.plot(x,y,label="SINQ", color="green")
plt.plot(x,z,label="COSQ",color="red")
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate values")
plt.legend(loc="upper left")
plt.show()
```

```python
plt.figure(figsize=(6,4))

plt.subplot(1,2,1)
plt.plot(x,y,label="SINQ", color="red")
plt.plot(x,z,label="COSQ",color="green")
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate values")
plt.legend()



plt.subplot(1,2,2)
plt.plot(x,y,label="SINQ", color="green")
plt.plot(x,z,label="COSQ",color="red")
plt.title("SINQ & COSQ")
plt.xlabel("Data Points")
plt.ylabel("Associate values")


#maintains a gap between to graphs w for width h for height
plt.subplots_adjust(wspace=0.5)

plt.legend()
plt.show()
```
