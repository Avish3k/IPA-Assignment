```python
try:
    a=int(input("Enter value for a"))
    b=int(input("Enter value for b"))
    c=a/b
    print(c)
except ZeroDivisionError:
    print("Division by zero is not allowed \n")
except ValueError:
    print("Enter only integer values")
```

```python
def safe_div():
    try:
        a=int(input("Enter value for a"))
        b=int(input("Enter value for b"))
        c=a/b
        print(c)
    except ZeroDivisionError:
        print("Division by zero is not allowed \n")
        safe_div()
    except ValueError:
        print("Enter only integer values")
        safe_div()

safe_div()
```

```python
#generic exception when exception is not known and error is not known
#keyboard interrupt is not handled in generix exception
try:
    a=int(input("Enter value for a"))
    b=int(input("Enter value for b"))
    c=a/b
    print(c)
except Exception as e:
    print("Exception is handled \n",e)
```

```python
#for keyboard interrupt we can use default try 
try:
    a=int(input("Enter value for a"))
    b=int(input("Enter value for b"))
    c=a/b
    print(c)
except ZeroDivisionError:
    print("Division by zero is not allowed \n")
except ValueError:
    print("Enter only integer values")  
else:
    print("Division Successful") 
finally: 
    print("Program executed")
```

```python
class AgeError(Exception):
    pass   # Custom exception

try:
    age = int(input("Enter age: "))
    
    if age < 18:
        raise AgeError("Age must be more than 18")
    else:
        print("Eligible")

except AgeError as e:
    print("Error:", e)

except ValueError:
    print("Please enter a valid number.")
```
